```python
from elasticsearch import Elasticsearch

es = Elasticsearch('http://localhost:9200')
es_info = es.info()
print(f"Connected to cluster: {es_info["cluster_name"]}")
print(f"Elasticsearch version: {es_info["version"]["number"]}")
```

# Crud Operations

### 1. create an index with mapping

```python
mapping = {
    "settings": {
        "number_of_shards": 1,
        "number_of_replicas":0
    },
    "mappings": {
        "properties": {
            "title": {"type": "text"},
            "author": {"type": "text"},
            "genre": {"type": "keyword"},
            "year": {"type": "integer"},
            "rating": {"type": "float"},
            "description": {"type": "text"},
            "tags": {"type": "keyword"},
            "published_at": {"type": "date", "format": "yyyy-MM-dd"}
        }
    }
}

index_name = "my-index"
if not es.indices.exists(index=index_name):
    response = es.indices.create(index=index_name, body=mapping)
    print(f"Index created : {response['acknowledged']}")
else:
    print(f"index with name {index_name} already exists!")
```

## Index  a single document (create an index)

```python
doc = {
    "title": "my-first-book",
    'author': 'venky',
    'rating': 6.9,
    'genre': 'comic',
    'description': 'feel good book with self development',
    'tags': ['book','venky','comic','self-help'],
    'year': 2002,
    'published_at': '2002-05-20'
    }

response = es.index(index=index_name, document=doc)
```

## Update element

### 1. partial update (recommended)

```python
partial_data  = {
    "rating": 8,
    "title": "updated title"
}
response = es.update(index=index_name, id="12", doc=partial_data)
```

```python
response = es.get(index=index_name, id="12")
pprint(response['_source'])
```

```python
# Full Replacement -> use index with same id

response = es.update(
    index=index_name,
    id="12",
    script={
        "source": f"ctx._source.rating += params.amount",
        "params": {"amount":12}
    }
)
```

```python
response = es.get(index=index_name, id="12")
pprint(response["_source"])
```
